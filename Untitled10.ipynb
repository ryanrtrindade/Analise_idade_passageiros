```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
base = pd.read_csv('/content/drive/MyDrive/PYTHON/train.csv')
```

```python
base2 = base.dropna(axis= 0, subset = 'Age')
```

```python
plt.figure(figsize = (20,10))

plt.title('Idade de passageiros que sobreviveram x morreram')
plt.xlabel('Idade')
plt.ylabel('Frequencia')
plt.hist(base2[base2.Survived == 1].Age, color = 'Blue', label = 'Sobreviveu', alpha = 0.5, bins = 30, edgecolor = 'Black')
plt.hist(base2[base2.Survived == 0].Age, color = 'RED', label = 'Não sobreviveu', alpha = 0.5, bins = 30, edgecolor = 'black' )
plt.legend(loc = 'upper right')
```

```python
plt.figure(figsize = (20,10))

media_sobreviventes = base2[base2.Survived == 1]['Age'].mean()
media_nao_sobreviventes = base2[base2.Survived == 0]['Age'].mean()

plt.bar('Sobreviveu', media_sobreviventes, color= 'Blue')
plt.bar('Não sobreviveu', media_nao_sobreviventes, color = 'Red')


plt.title('Média de idade dos sobreviventes x não sobreviventes')
plt.xlabel('Sobreviventes')
plt.ylabel('Média de sobreviventes')
plt.legend(loc = 'upper right')

```
